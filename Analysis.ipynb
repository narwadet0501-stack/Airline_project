```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\SWATI\Desktop\Internship Projects\Final_Project\Airline_project\Airline1.csv")
```

```python
# Find Missing Values and fill them
```

```python
df.isnull().sum()
```

```python
df[df['Total_Stops'].isnull()]
```

```python
Missing_Total_Stops = df['Total_Stops'].mode()[0]
print(Missing_Total_Stops)
df.fillna({'Total_Stops':Missing_Total_Stops},inplace=True)
```

```python
Missing_Destination = df['Destination'].mode()[0]
df.fillna({'Destination':Missing_Destination},inplace=True)
```

```python
df.head()
```

```python
df[df['Source'].isnull()]
```

```python
Missing_Route = df['Route'].mode()[0]
df.fillna({"Route":Missing_Route},inplace=True)
```

```python
df.isnull().sum()

Missing_Price = df['Price'].mean()
Missing_Price

df.fillna({'Price':Missing_Price},inplace=True)
```

```python
df.isnull().sum()
```

```python
# Find duplicate values
df[df.duplicated()]
```

```python
# Remove the duplicate values
df.drop_duplicates(inplace=True)
```

Graph Based Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Nedd to sort values first for sorting a graph
AirCount = df.groupby('Airline')['Airline'].count().sort_values()
AirCount.index
```

```python
# Distribution of Airlines

ax = sns.countplot(x = df['Airline'], order= AirCount.index)
ax.bar_label(ax.containers[0])
plt.xticks(rotation=45)
plt.title('Airline')
plt.savefig("Presentation/20.Distrubution_of_Airline.png")
plt.show()
```

```python
# Distribution of Source

Source_count = df.groupby('Source')['Source'].count().sort_values() #(Data Sorting)

ax = sns.countplot(x = df['Source'], order=Source_count.index)
ax.bar_label(ax.containers[0])

plt.title('Source')
plt.savefig("Presentation/1.Distrubution_of_Source.png")
plt.show()
```

```python
# Airline against Price

Air_Price = df.groupby('Airline')['Price'].sum().sort_values()

ax = sns.barplot(x=df['Airline'],y=df['Price'],estimator='sum',order=Air_Price.index,errorbar=None)
ax.bar_label(ax.containers[0],rotation=50,fmt="%.0f")
plt.xticks(rotation=90)

plt.title('Airline against Price')
plt.savefig("Presentation/2.Airline_against_Price.png")
plt.show()
```

```python
# Source against Price

Source_Price = df.groupby('Source')['Price'].sum().sort_values() #(data Sorting)

ax = sns.barplot(x=df['Source'],y=df['Price'],estimator='sum',errorbar=None,order = Source_Price.index)
ax.bar_label(ax.containers[0],fmt="%.0f")

plt.title('Source against Price')
plt.savefig("Presentation/3.Source_against_Price.png")
plt.show()
```

```python
# Total_Stops against Price

Total_Stops_Price = df.groupby('Total_Stops')['Price'].sum().sort_values() #(Sorting)

ax = sns.barplot(x = df['Total_Stops'],y=df['Price'],estimator='sum',order=Total_Stops_Price.index,errorbar=None)
ax.bar_label(ax.containers[0],fmt="%.0f")

plt.title('Total_Stops against Price')
plt.savefig("Presentation/4.TotalStops_against_Price.png")
plt.show()
```

```python
# Avg Total_stops Count for Price

TtlDf = df.groupby('Total_Stops')['Price'].agg(['mean','count']).sort_values(by='mean')
TtlDf

ax = sns.barplot(TtlDf['mean'])
ax.bar_label(ax.containers[0])

ax.twinx()
ax2 = sns.lineplot(TtlDf['count'],color='orange',marker="o")

plt.title('Avg Total_stops price for airline')
plt.savefig("Presentation/5.AvgTotalStopPriceForAirline.png")
plt.show()
```

```python
#  AVG PRICE COUNT for Airline

Air_Avg = df.groupby('Airline')['Price'].agg(['mean','count']).sort_values(by='mean')
Air_Avg

ax = sns.barplot(Air_Avg['mean'])
ax.bar_label(ax.containers[0],rotation=90)
plt.xticks(rotation=90)
ax.twinx()
ax2 = sns.lineplot(Air_Avg['count'],color='orange',marker="o")

plt.savefig("Presentation/6.AvgPriceCountforAirline.png")
plt.show()
```

```python
# Min PRICE vs COUNT for Airline

Air_Min = df.groupby('Airline')['Price'].agg(['mean','count']).sort_values(by='mean')
Air_Min

ax = sns.barplot(Air_Min['mean'])
ax.bar_label(ax.containers[0],rotation=90)
plt.xticks(rotation=90)
ax.twinx()
ax2 = sns.lineplot(Air_Min['count'],color='orange',marker="o")

plt.savefig("Presentation/7.AVGPRICECOUNTforAirline.png")
plt.show()
```

```python
df['Sector'] = df['Source']+" - "+df['Destination']

stops_df = df[df['Total_Stops']=='non-stop']
Sector_Df = stops_df.groupby('Sector')['Price'].agg('mean')

ax = sns.barplot(Sector_Df)
ax.bar_label(ax.containers[0])
plt.xticks(rotation=45)
plt.title('non-stop')
plt.savefig("Presentation/9.TotalStops_non-stop.png")
plt.show()
```

```python
df['Sector'] = df['Source']+" - "+df['Destination']
list_Stops = enumerate(['non-stop','1 stop','2 stops','3 stops','4 stops'])
plt.figure(figsize=(15,20))
for index, value in list_Stops:
    Stops_Df = df[df['Total_Stops']==value]
    
    Sector_Df = Stops_Df.groupby('Sector')['Price'].agg('mean')
    plt.subplot(3,2,index+1)
    ax = sns.barplot(Sector_Df)
    
    ax.bar_label(ax.containers[0])
    plt.title(value)

plt.savefig("Presentation/8.Avg_FareOfEachSectorWithTotal-stops.png")
plt.show()
```
